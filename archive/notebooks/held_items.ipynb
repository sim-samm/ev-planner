## Stage 2 Design: Macho Brace

Let:

- `B` be battles needed
- `R` be remaining EVs needed for the trainee stat
- `Y` be total EV yield from the battle opponents list
- `M` be the Macho Brace multiplier (commonly `2`)

Baseline (Stage 1):

$$
B = \left\lceil \frac{R}{Y} \right\rceil
$$

With Macho Brace multiplying EV yield:

$$
B_{\text{macho}} = \left\lceil \frac{R}{M \cdot Y} \right\rceil
$$

Expected checkpoint for Geodude (`Y=1`, `M=2`, `R=252`):

$$
B_{\text{macho}} = \left\lceil \frac{252}{2} \right\rceil = 126
$$

```python
# Simulate a trainee with a held item
# TODO: trainee object should have a held item

macho_brace = {
    "item": "Macho Brace",
    "ev_multiplier": 2,
    "speed_multiplier": 0.5, # future operation-cost consideration; not used for EV math
}

held_item = macho_brace;
```

```python
# Simulate battle with a held item
# TODO: these functions seem like they should be combined

# TODO: maybe rename the ev_from_battle function to ev_from_battle_baseline

def ev_from_battle_with_held_item(opponents, held_item):
    return ev_from_battle(opponents) * held_item["ev_multiplier"]

# TODO: maybe rename the battles_needed function to battles_needed_baseline

def battles_needed_with_held_item(trainee, opponents, held_item):
    remaining_evs_needed = trainee["target_evs"] - trainee["current_evs"]
    ev_gain = ev_from_battle_with_held_item(opponents, held_item)
    return math.ceil(remaining_evs_needed / ev_gain)
```

```python
# Run the  held item battle

held_item_battle = ev_from_battle_with_held_item(opponents, held_item)
held_item_battles = battles_needed_with_held_item(trainee, opponents, held_item)

print(f"Held item battle EV gain: {held_item_battle}")
print(f"Held item battles needed: {held_item_battles}")
```
