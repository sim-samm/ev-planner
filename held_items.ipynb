## Stage 2 Design: Macho Brace

Let:

- `B` be battles needed
- `R` be remaining EVs needed for the trainee stat
- `Y` be total EV yield from the battle opponents list
- `M` be the Macho Brace multiplier (commonly `2`)

Baseline (Stage 1):

$$
B = \left\lceil \frac{R}{Y} \right\rceil
$$

With Macho Brace multiplying EV yield:

$$
B_{\text{macho}} = \left\lceil \frac{R}{M \cdot Y} \right\rceil
$$

Expected checkpoint for Geodude (`Y=1`, `M=2`, `R=252`):

$$
B_{\text{macho}} = \left\lceil \frac{252}{2} \right\rceil = 126
$$
