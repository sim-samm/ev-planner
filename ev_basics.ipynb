# EV Basics: Pokemon Training Plan

This notebook models the Stage 1 baseline: one trainee targeting one stat, with basic EV math from battle yields.

## Scope

- Includes simple EV yield math and battles-needed calculations.
- Excludes advanced mechanics for now.
- For upcoming stages, see `docs/roadmap.md`.

```python
import math
```

## Trainee Input

```python
trainee = {
    "species": "Magikarp",
    "target_stat": "speed",
    "current_evs": 0,
    "target_evs": 252,
}
```

## Opponent Pool


```python
opponents = [
    {
        "species": "Geodude",
        "ev_yield_stat": "defense",
        "ev_yield": 1,
    },
    {
        "species": "Graveler",
        "ev_yield_stat": "defense",
        "ev_yield": 2,
    },
]
```

## Core EV Math

```python
def ev_from_battle(opponents):
    entries = [opponents] if isinstance(opponents, dict) else list(opponents)
    return sum(opp["ev_yield"] for opp in entries)
```

## Checkpoint: Verify `ev_from_battle`

Quick sanity checks before building the next function.

```python
print("Single opponent EV gain:", ev_from_battle(opponents[0]))
print("Combined opponent EV gain:", ev_from_battle(opponents))
```

## Battles Needed Function

```python
def battles_needed(trainee, opponents):
    remaining_evs_needed = trainee["target_evs"] - trainee["current_evs"]
    ev_gain = ev_from_battle(opponents)
    return math.ceil(remaining_evs_needed / ev_gain)

print("Geodude battles needed:", battles_needed(trainee, opponents[0]))
print("Graveler battles needed:", battles_needed(trainee, opponents[1]))
```

## Build Battle Plan Rows

```python
def battle_plan_rows(trainee, opponents):
    rows = []
    for opp in opponents:
        rows.append({
            "species": opp["species"],
            "battles_needed": battles_needed(trainee, opp),
        })
    return rows

rows = battle_plan_rows(trainee, opponents)
for row in sorted(rows, key=lambda r: r["battles_needed"]):
    print(f"Battle {row['species']}: {row['battles_needed']}")
```
