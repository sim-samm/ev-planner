# Single Pokemon Training Plan 
EV planning concept for training one pokemon against one opponent.

```python
# A simple data structure for the pokemon being trained.

trainee = {
    "species": "Magikarp",
    "target_stat": "speed",
    "current_evs": 0,
    "target_evs": 252,
}
```

```python
# A simple data structure for the opponent being fought.

opponent = {
    "species": "Geodude",
    "ev_yield_stat": "defense",
    "ev_yield": 1,
}
```

```python
# EVs from battle calculation
# TODO: need another method for this
# def ev_from_battle. ars: trainer or wild. returns: ev_gain

def ev_from_battle(opponent_list):
    # Normalize input: dict -> [dict], list/iterable -> list
    opponents = [opponent_list] if isinstance(opponent_list, dict) else list(opponent_list)
    total = 0
    for opp in opponents:
        total += opp["ev_yield"]
    return total
```

```python
# Calculate the Battles Needed
import math

def battles_needed(trainee, opponent_list):
    remaining_evs_needed = trainee["target_evs"] - trainee["current_evs"]
    ev_gain = ev_from_battle(opponent_list)
    return math.ceil(remaining_evs_needed / ev_gain)

# Print the results
needed = battles_needed(trainee, opponent)
print(f"Battles needed: {needed}")
```

```python
# Second Opponent
opponent_b = {
    "species": "Graveler",
    "ev_yield_stat": "defense",
    "ev_yield": 2,
}
```

```python
# Opponent list 
# TODO: maybe onther structure will be better later
opponent_list = []
opponent_list.append(opponent)
opponent_list.append(opponent_b)
print(opponent_list)
```

```python
# Compute the EVs needed per battle
def list_battle_ev_totals(opponent_list):
    ev_totals = []
    for opp in opponent_list:
        opp_ev_total = {
            "opponent": opp["species"],
            "ev_totals": battles_needed(trainee, opp)
        }
        ev_totals.append(opp_ev_total)
    return ev_totals
```

```python
battle_ev_totals = list_battle_ev_totals(opponent_list)

for battle in battle_ev_totals:
    print(f"Battle {battle['opponent']}: {battle['ev_totals']}")
```
